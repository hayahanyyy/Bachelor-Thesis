```python
import pandas as pd
import numpy as np


# 2. Data Cleaning
# Check for missing data and duplicating data
print(f"Initial row count: {len(df)}")
df.drop_duplicates(inplace=True)
df.dropna(inplace=True)
print(f"Row count after cleaning: {len(df)}")

# 3. Data Transformation
# Task: Replacing numerical values with names (Example for Marital Status)
# The paper mentions replacing codes with descriptive names for better analysis
marital_status_map = {1: 'Single', 2: 'Married', 3: 'Widower', 4: 'Divorced', 5: 'Facto Union', 6: 'Legally Separated'}
df['Marital status'] = df['Marital status'].map(marital_status_map)

# Task: Changing categorical columns to 'category' data type
# The paper identifies specific categorical categories like Gender, Nationality, and Course [cite: 428]
categorical_cols = ['Marital status', 'Application mode', 'Course', 'Daytime/evening attendance',
                    'Previous qualification', 'Nacionality', "Mother's qualification",
                    "Father's qualification", "Mother's occupation", "Father's occupation",
                    'Displaced', 'Educational special needs', 'Debtor', 'Tuition fees up to date',
                    'Gender', 'Scholarship holder', 'International']

for col in categorical_cols:
    df[col] = df[col].astype('category')

# 4. Feature Selection
# The paper selected attributes with a direct impact on student status
# Top factors include Curricular units, Tuition fees, and Scholarship status [cite: 172]
selected_features = [
    'Curricular units 2nd sem (approved)', 'Curricular units 2nd sem (grade)',
    'Curricular units 1st sem (approved)', 'Curricular units 1st sem (grade)',
    'Tuition fees up to date', 'Scholarship holder', 'Age at enrollment',
    'Debtor', 'Gender', 'Application mode'
]

# Ensure the target variable is mapped to the categories used in the paper
# Target attributes: graduate, dropout, and enrolled
df['Target'] = df['Target'].astype('category')

# 5. Final Preparation for Modeling
# Get all categorical variables except target for the rule engine/analysis
categorical_features = df.select_dtypes(include=['category']).columns.drop('Target')

print("Data Preparation Phase Complete.")
print(df.info())
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score, GridSearchCV
```

```python
# 1. Initialize the 10-fold Stratified K-Fold
# This ensures that each fold has the same percentage of samples of each target class
skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)
```

```python
rf_model = RandomForestClassifier(n_estimators=100, random_state=42)
```

2. Example: Training the Best Model (Random Forest)




```python
# 2. Perform the 80/20 split as per the research method (Fig. 2)
# We use stratify=y to ensure the class distribution remains consistent in both sets
x_train, x_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# 3. Initialize the 10-fold Stratified K-Fold for later model evaluation
skf = StratifiedKFold(n_splits=10, shuffle=True, random_state=42)

print(f"Training set size: {len(x_train)}")
print(f"Testing set size: {len(x_test)}")
```

```python


# 3. Use GridSearchCV for hyperparameter tuning as mentioned in the paper
param_grid = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 10, 20]
}
grid_search = GridSearchCV(estimator=rf_model, param_grid=param_grid, cv=skf, scoring='accuracy')
grid_search.fit(x_train, y_train)

best_rf = grid_search.best_estimator_
print(f"Best Accuracy from GridSearch: {grid_search.best_score_:.4f}")
```
